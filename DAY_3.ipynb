```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

np.random.seed(42)
print('Setup complete. NumPy', np.__version__)
```

```python
# P(A) = favourable outcomes / total outcomes.
# We can estimate it by simulating many trials.
rolls = np.random.randint(1, 7, size=100_000)   # 100k dice rolls

p_even = (rolls % 2 == 0).mean()       # P(even)
p_gt4  = (rolls > 4).mean()            # P(roll > 4)
print('P(even) ~', round(p_even, 3), ' (true 0.5)')
print('P(>4)   ~', round(p_gt4, 3),  ' (true 0.333)')
```

```python
p_1 = (rolls == 1).mean()
p_2 = (rolls == 2).mean()
p_1_or_2 = (np.isin(rolls, [1, 2])).mean()
print('P(1) + P(2) =', round(p_1 + p_2, 3))
print('P(1 or 2)   =', round(p_1_or_2, 3), ' -> they match')
```

```python
import numpy as np

# Simulate 100,000 pairs of coin flips
flips = np.random.randint(0, 2, size=(100_000, 2))
heads = flips.sum(axis=1)   # number of heads in each pair

# 1. P(both heads)
p_both_heads = (heads == 2).mean()
print("P(both heads) =", round(p_both_heads, 3))

# 2. P(at least one head)
p_at_least_one = (heads >= 1).mean()
print("P(at least one head) =", round(p_at_least_one, 3))

# 3. Check P(0) + P(1) + P(2)
p0 = (heads == 0).mean()
p1 = (heads == 1).mean()
p2 = (heads == 2).mean()

print("P(0 heads) =", round(p0, 3))
print("P(1 head)  =", round(p1, 3))
print("P(2 heads) =", round(p2, 3))
print("Sum =", round(p0 + p1 + p2, 3))
```

```python
rolls = np.random.randint(1, 7, size=100_000)

# P(roll is 6 | roll is even):  narrow the world to even rolls
even = rolls[rolls % 2 == 0]          # condition on 'even'
p_6_given_even = (even == 6).mean()
print('P(6 | even) ~', round(p_6_given_even, 3), ' (true 1/3)')
```

```python
A = rolls > 3
B = rolls % 2 == 0
p_A        = A.mean()
p_A_given_B = A[B].mean()
print('P(A)      =', round(p_A, 3))
print('P(A | B)  =', round(p_A_given_B, 3))
print('Independent?', np.isclose(p_A, p_A_given_B, atol=0.02))
```

```python
import pandas as pd
import numpy as np

print("\n--- Task 2: Synthetic Data Generation for Practice ---")

# Create DataFrame
df_sales = pd.DataFrame({
    'Date': pd.date_range(start='2023-01-01', end='2023-04-10', periods=100),
    'Region': np.random.choice(
        ['East', 'West', 'North', 'South'],
        size=100
    ),
    'Product': np.random.choice(
        ['Laptop', 'Mouse', 'Keyboard', 'Monitor'],
        size=100
    ),
    'Sales': np.random.randint(100, 1001, size=100),
    'Quantity': np.random.randint(1, 11, size=100)
})

# Introduce 5 NaN values in Sales
sales_nan_idx = np.random.choice(df_sales.index, size=5, replace=False)
df_sales.loc[sales_nan_idx, 'Sales'] = np.nan

# Introduce 3 NaN values in Quantity
qty_nan_idx = np.random.choice(df_sales.index, size=3, replace=False)
df_sales.loc[qty_nan_idx, 'Quantity'] = np.nan

# Display first few rows
print(df_sales.head())

# Verify missing values
print("\nMissing Values:")
print(df_sales.isna().sum())
```

```python
print("\n--- Task 3: Indexing and Slicing in DataFrames ---")

# 3.1 Select only 'Region' and 'Sales' columns
df3_1 = df_sales[['Region', 'Sales']]

print("\n3.1 Region and Sales:")
print(df3_1.head())

# 3.2 Select rows where Product is Laptop
df3_2 = df_sales[df_sales['Product'] == 'Laptop']

print("\n3.2 Rows where Product = Laptop:")
print(df3_2.head())

# 3.3 Region = East AND Sales > 500
df3_3 = df_sales[
    (df_sales['Region'] == 'East') &
    (df_sales['Sales'] > 500)
]

print("\n3.3 East region and Sales > 500:")
print(df3_3.head())

# 3.4 Sales value for the first Monitor entry
val3_4 = df_sales.loc[
    df_sales['Product'] == 'Monitor',
    'Sales'
].iloc[0]

print("\n3.4 First Monitor Sales Value:")
print(val3_4)
```

```python
print("\n--- Task 4: Data Cleaning (Handling Missing Values) ---")

# 4.1 Count missing values in each column
missing_counts4_1 = df_sales.isna().sum()

print("\nMissing Values Per Column:")
print(missing_counts4_1)

# 4.2 Fill missing Sales values with the mean Sales
df4_2_filled = df_sales.copy()

sales_mean = df4_2_filled['Sales'].mean()

df4_2_filled['Sales'] = df4_2_filled['Sales'].fillna(sales_mean)

print("\nMissing Sales after filling:")
print(df4_2_filled['Sales'].isna().sum())

# 4.3 Drop rows containing any missing values
df4_3_dropped = df_sales.dropna()

print("\nOriginal Shape :", df_sales.shape)
print("After dropna() :", df4_3_dropped.shape)
```

```python
print("\n--- Task 5: Data Manipulation (Columns, Sorting, Grouping) ---")

# 5.1 Create Total_Revenue column
df_sales['Total_Revenue'] = df_sales['Sales'] * df_sales['Quantity']

print("\nTotal_Revenue column added:")
print(df_sales[['Sales', 'Quantity', 'Total_Revenue']].head())

# 5.2 Sort by Date (ascending) then Sales (descending)
df5_2_sorted = df_sales.sort_values(
    by=['Date', 'Sales'],
    ascending=[True, False]
)

print("\nSorted Data:")
print(df5_2_sorted.head())

# 5.3 Group by Region and calculate sum of Sales and Quantity
df5_3_grouped_region = df_sales.groupby('Region')[
    ['Sales', 'Quantity']
].sum()

print("\nSales and Quantity Sum by Region:")
print(df5_3_grouped_region)

# 5.4 Group by Product and calculate average Sales
df5_4_grouped_product = df_sales.groupby('Product')[
    'Sales'
].mean()

print("\nAverage Sales by Product:")
print(df5_4_grouped_product)
```
