```python
import random
from datetime import datetime, timedelta

# Constants
DEVICE_TYPES = ["Temperature", "Humidity", "Pressure", "Light"]
LOCATIONS = ["Room_A", "Room_B", "Lab_1", "Lab_2", "Warehouse"]

READING_RANGES = {
    "Temperature": (15.0, 40.0),  # °C
    "Humidity": (20.0, 90.0),     # %
    "Pressure": (950.0, 1050.0),  # hPa
    "Light": (100.0, 1000.0)      # Lux
}

# Generate simulated sensor readings
raw_sensor_data = []

start_time = datetime.now()

for i in range(1, 101):
    device_id = f"DEV{i:03d}"
    device_type = random.choice(DEVICE_TYPES)
    location = random.choice(LOCATIONS)

    min_val, max_val = READING_RANGES[device_type]
    value = round(random.uniform(min_val, max_val), 2)

    timestamp = start_time + timedelta(seconds=i)
    timestamp_str = timestamp.strftime("%Y-%m-%d %H:%M:%S")

    raw_sensor_data.append(
        (timestamp_str, device_id, device_type, location, value)
    )

# Example: Print first 5 readings
for reading in raw_sensor_data[:99]:
    print(reading)
```

```python
temperature_readings = []

for reading in raw_sensor_data:
    timestamp, device_id, device_type, location, value = reading

    if device_type == "Temperature_Sensor":
        temp_dict = {
            "timestamp": timestamp,
            "device_id": device_id,
            "location": location,
            "temperature_celsius": value
        }
        temperature_readings.append(temp_dict)

# Print first 5 records
print(temperature_readings[:5])
```

```python
unique_locations = set()

for reading in raw_sensor_data:
    unique_locations.add(reading[3])

print(unique_locations)
```

```python
import numpy as np

s1 = np.array([85, 90, 78])
s2 = np.array([92, 88, 95])
s3 = np.array([76, 81, 84])

# L2 norms
print("L2 Norms:")
for i, s in enumerate([s1, s2, s3], start=1):
    print(f"Student {i}: {np.linalg.norm(s):.2f}")

# Cosine similarity
def cosine_similarity(a, b):
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))

print("\nCosine Similarities:")
print("S1-S2:", cosine_similarity(s1, s2))
print("S1-S3:", cosine_similarity(s1, s3))
print("S2-S3:", cosine_similarity(s2, s3))
```

```python
import numpy as np

# Coefficient matrix
A = np.array([
    [2, 3, 1],
    [1, 2, 4],
    [3, 1, 2]
])

# Constants vector
B = np.array([110, 90, 130])

# Solve the system
X = np.linalg.solve(A, B)

print("Solution [x, y, z]:")
print(X)

# Verification
print("\nVerification (A @ X):")
print(A @ X)

print("\nOriginal B:")
print(B)

print("\nVerification passed:",
      np.allclose(A @ X, B))
```
