```python
import numpy as np
import numpy.linalg as la

# Cosine similarity function
def cosine(u, v):
    return np.dot(u, v) / (la.norm(u) * la.norm(v))

pairs = [
    (np.array([1, 0, 1]), np.array([1, 0, 1])),    # pair 1
    (np.array([1, 2, 3]), np.array([3, 2, 1])),    # pair 2
    (np.array([2, 0, 0]), np.array([0, 5, 0])),    # pair 3
]

similarities = []

for i, (u, v) in enumerate(pairs, start=1):
    l1 = la.norm(u, 1)      # L1 norm
    l2 = la.norm(u)         # L2 norm
    cos_sim = cosine(u, v)  # Cosine similarity

    similarities.append(cos_sim)

    print(f"\nPair {i}")
    print("u =", u)
    print("v =", v)
    print("L1 norm of u :", l1)
    print("L2 norm of u :", round(l2, 3))
    print("Cosine similarity :", round(cos_sim, 3))

# Most similar pair
most_similar = np.argmax(similarities) + 1
print(f"\nMost similar pair = Pair {most_similar}")
```

```python
 A= np.array([[2., 1.],
              [1., 3.]])
B = np.array([[1., 0.],
              [4., 2.]])

print('A @ B (matrix multiply):\n', A @ B)
print('A.T   (transpose):\n', A.T)
print('inverse(A):\n', la.inv(A))
print('trace(A) = sum of diagonal:', np.trace(A))
```

```python
I = np.eye(2)                       # identity: 1s on the diagonal
print('Identity I:\n', I)
print('A @ inv(A) == I ?', np.allclose(A @ la.inv(A), I))

# Symmetric: A equals its own transpose
print('A symmetric?', np.allclose(A, A.T))

# Orthogonal: Q.T @ Q == I (a pure rotation/reflection)
theta = np.radians(30)
Q = np.array([[np.cos(theta), -np.sin(theta)],
              [np.sin(theta),  np.cos(theta)]])
print('Q orthogonal?', np.allclose(Q.T @ Q, I))
```

```python
import numpy as np
square = np.array([[0, 1, 1, 0],
                   [0, 0, 1, 1]], dtype=float)

# Scaling matrix: stretch x by 1.5, y by 0.5
S = np.array([[1.5, 0.0],
              [0.0, 0.5]])

# Rotation matrix: rotate by 30 degrees
t = np.radians(30)
R = np.array([[np.cos(t), -np.sin(t)],
              [np.sin(t),  np.cos(t)]])

scaled  = S @ square      # apply scaling
rotated = R @ square      # apply rotation
print('Rotated corners:\n', rotated)
```

```python
import matplotlib.pyplot as plt
def close_loop(pts):
    # repeat the first point at the end so the polygon closes
    return np.hstack([pts, pts[:, :1]])

fig, ax = plt.subplots(figsize=(5, 5))
ax.plot(*close_loop(square),  marker='o', label='original')
ax.plot(*close_loop(scaled),  marker='o', label='scaled')
ax.plot(*close_loop(rotated), marker='o', label='rotated 30 deg')
ax.axhline(0, color='gray', lw=0.5); ax.axvline(0, color='gray', lw=0.5)
ax.set_aspect('equal'); ax.legend(); ax.set_title('Linear transformations')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Triangle vertices
tri = np.array([[0, 1, 0.5],
                [0, 0, 1.0]])

# 1. Scaling matrix (x*2, y*0.5)
S = np.array([[2.0, 0.0],
              [0.0, 0.5]])

# 2. Rotation matrix (45 degrees)
theta = np.radians(45)

R = np.array([[np.cos(theta), -np.sin(theta)],
              [np.sin(theta),  np.cos(theta)]])

# 3. Apply scaling first, then rotation
transformed = R @ (S @ tri)

# Function to close the triangle loop
def close_loop(points):
    return np.hstack([points, points[:, [0]]])

# Close both shapes for plotting
tri_closed = close_loop(tri)
trans_closed = close_loop(transformed)

# Plot
plt.figure(figsize=(6,6))

plt.plot(tri_closed[0], tri_closed[1],
         marker='o', label='Original Triangle')

plt.plot(trans_closed[0], trans_closed[1],
         marker='s', label='Scaled + Rotated Triangle')

plt.axis('equal')
plt.grid(True)
plt.legend()
plt.title("Original vs Transformed Triangle")

plt.show()
```

```python
import numpy as np
import numpy.linalg as la

# Matrix A
A = np.array([[2., 0.],
              [0., 3.]])

# Compute eigenvalues and eigenvectors
vals, vecs = la.eig(A)

print("Matrix A:\n", A)
print("\nEigenvalues (lambda):")
print(vals)

print("\nEigenvectors (columns):")
print(vecs)

# Verify Av = lambda*v for each eigenvector
for i in range(len(vals)):
    lam = vals[i]        # eigenvalue
    v = vecs[:, i]       # corresponding eigenvector

    print(f"\nEigenpair {i+1}")
    print("lambda =", lam)
    print("v =", v)

    print("A @ v =")
    print(A @ v)

    print("lambda * v =")
    print(lam * v)

    print("Verification:",
          np.allclose(A @ v, lam * v))
```

```python

for i in range(len(vals)):
    v = vecs[:, i]                 # i-th eigenvector
    lhs = A @ v                    # A v
    rhs = vals[i] * v              # lambda v
    print(f'lambda={vals[i]:.1f}  A v == lambda v ?', np.allclose(lhs, rhs))
```

```python
import numpy as np
import numpy.linalg as la

C = np.array([[4., 1.],
              [2., 3.]])

# 1. Eigenvalues and eigenvectors
vals, vecs = la.eig(C)

# 2. Print eigenvalues
print("Eigenvalues:")
print(vals)

print("\nEigenvectors (columns):")
print(vecs)

# 3. Verify C @ v = lambda * v for the first eigenvector
lam = vals[0]      # first eigenvalue
v = vecs[:, 0]     # first eigenvector

print("\nFirst eigenvalue (lambda):", lam)
print("First eigenvector (v):", v)

print("\nC @ v:")
print(C @ v)

print("\nlambda * v:")
print(lam * v)

print("\nVerification:",
      np.allclose(C @ v, lam * v))
```

```python
A = np.array([[ 2.,  1., -1.],
              [-3., -1.,  2.],
              [-2.,  1.,  2.]])
b = np.array([8., -11., -3.])

x = la.solve(A, b)
print('Solution x:', x)                    # [2, 3, -1]
print('Rank of A :', la.matrix_rank(A))     # 3 -> full rank
print('Full rank -> a unique solution exists')
```

```python
import numpy as np
import numpy.linalg as la

A2 = np.array([[3., 2.],
               [1., 4.]])

b2 = np.array([7., 9.])

cat = np.array([0.9, 0.8, 0.1])
dog = np.array([0.85, 0.7, 0.2])
car = np.array([0.1, 0.2, 0.95])

# Cosine similarity function
def cosine(u, v):
    return np.dot(u, v) / (la.norm(u) * la.norm(v))

# 1. Solve A2 x = b2
x = la.solve(A2, b2)
print("Solution x:")
print(x)

# 2. Rank of A2
print("\nRank of A2:")
print(la.matrix_rank(A2))

# 3. Cosine similarities
cat_dog = cosine(cat, dog)
cat_car = cosine(cat, car)

print("\nCosine(cat, dog) =", round(cat_dog, 3))
print("Cosine(cat, car) =", round(cat_car, 3))

if cat_dog > cat_car:
    print("\ncat and dog are more similar.")
else:
    print("\ncat and car are more similar.")
```
